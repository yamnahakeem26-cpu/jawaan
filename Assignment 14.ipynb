```python
import pandas as pd
import sqlite3 as sql
```

```python
conn = sql.connect('database.sqlite')
conn
```

```python
pd.read_sql_query('select * from sqlite_master',conn)
```

```python
pd.read_sql_query('select * from salaries',conn)
```

```python
def sql_query(q):
    return pd.read_sql_query(q,conn)
    
```

#### **1. What is the average TotalPayBenefits for each JobTitle? exclude ('Not Provided', 'Not provided')**

```python
sql_query("""SELECT JobTitle, AVG(TotalPayBenefits) AS AveragePay
FROM salaries
WHERE JobTitle NOT IN ('Not Provided', 'Not provided')
GROUP BY JobTitle""")
```

#### **2. Show The JobTitle and Overtime Pay for All Employees With Overtime Pay greater than 50000**

```python
sql_query("""select jobtitle , overtimepay  from salaries 
         where overtimepay >50000 """ )
```

#### **3. What is the total number of unique employees? exclude ('Not Provided', 'Not provided')**

```python
sql_query('select count(distinct(employeename)) as uniqueemployeename from salaries where employeename not in ("Not Provided" , "Not provided")')
```

#### **4.Show The Top 10 highest Paid Employees**

```python
sql_query('''SELECT employeename,totalpaybenefits
FROM salaries
ORDER BY TotalPayBenefits desc
LIMIT 10''')
```

#### **5.Show All Employees Ordered By Their TotalPay Benefits In Descending Order**

```python
sql_query('''SELECT *
from salaries
ORDER BY TotalPayBenefits desc''')
```
